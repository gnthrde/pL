```python
# MIT License
# Copyright (c) 2026 T.H.Günther, mail@gnthr.de

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-1, 5, 1000)

a1=0.7
a2=1
a3=1.4


y1 = a1 / (1 + a1**2*x**2)
y2 = a2 / (1 + a2**2*x**2)
y3 = a3 / (1 + a3**2*x**2)

fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(x, y1, label=r'$a=0.7$', color='#33b0ac', linewidth=2, linestyle=':')
ax.plot(x, y2, label=r'$ a=1.0 $', color='#2a89a8', linewidth=2, linestyle='--')
ax.plot(x, y3, label=r'$ a=1.4 $', color='#341c4d', linewidth=2)

ax.axhline(0, color='black', linewidth=0.8, linestyle='--')  
ax.axvline(0, color='black', linewidth=0.8, linestyle='--')  
ax.set_xlabel('r', fontsize=12)
ax.set_ylabel('$ R_a(r) $', fontsize=12)
ax.legend(fontsize=11)

ax.text(-0.12, 0.99, 'C', transform=ax.transAxes, 
        fontsize=14, fontweight='bold', va='top', ha='left')


plt.savefig("PL_ArealR_const-a_slice.png", dpi=300, bbox_inches="tight")

plt.show()
```
