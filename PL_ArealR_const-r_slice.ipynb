```python
# MIT License
# Copyright (c) 2026 T.H.Günther, mail@gnthr.de

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 1000)

y1 = x / (1 + 0.81*x**2)
y2 = x / (1 + x**2)
y3 = x / (1 + 1.44*x**2)

fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(x, y1, label=r'$r=0.9$', color='#33b0ac', linewidth=2, linestyle=':')
ax.plot(x, y2, label=r'$ r=1.0 $', color='#2a89a8', linewidth=2, linestyle='--')
ax.plot(x, y3, label=r'$ r=1.2 $', color='#341c4d', linewidth=2)

ax.axhline(0, color='black', linewidth=0.8, linestyle='--')  
ax.axvline(0, color='black', linewidth=0.8, linestyle='--')  
ax.set_xlabel('a', fontsize=12)
ax.set_ylabel('$ R_r(a) $', fontsize=12)
ax.legend(fontsize=11)

ax.text(-0.12, 0.99, 'A', transform=ax.transAxes, 
        fontsize=14, fontweight='bold', va='top', ha='left')


plt.savefig("PL_ArealR_const-r_slice.png", dpi=300, bbox_inches="tight")

plt.show()
```
