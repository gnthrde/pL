```python
# MIT License
# Copyright (c) 2026 T.H.Günther, mail@gnthr.de

import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(-1, 4, 1000)
x=np.exp(t)

y1 = x / (1 + 0.81*x**2)
y2 = x / (1 + x**2)
y3 = x / (1 + 1.44*x**2)

fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(t, y1, label=r'$ r=0.9 $', color='#33b0ac', linewidth=2, linestyle=':')
ax.plot(t, y2, label=r'$ r=1.0 $', color='#2a89a8', linewidth=2, linestyle='--')
ax.plot(t, y3, label=r'$ r=1.2 $', color='#341c4d', linewidth=2)

ax.axhline(0, color='black', linewidth=0.8, linestyle='--')  
ax.axvline(0, color='black', linewidth=0.8, linestyle='--')  
ax.set_xlabel('time t', fontsize=12)
ax.set_ylabel('$ R_r(t) $  $ (vacuum) $', fontsize=12)
ax.legend(fontsize=11)

ax.text(-0.12, 0.99, 'B', transform=ax.transAxes, 
        fontsize=14, fontweight='bold', va='top', ha='left')

plt.savefig("PL_ArealR_vacuum-const-r_slice.png", dpi=300, bbox_inches="tight")
plt.show()


```
